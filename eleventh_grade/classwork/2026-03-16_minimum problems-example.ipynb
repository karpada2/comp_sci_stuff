```python
import numpy as np
```

## Example

```python
target = 18.73

best_number = None
best_distance = float("inf")

for x in range(51):  
    distance = abs(x - target)
    if distance < best_distance:
        best_distance = distance
        best_number = x

print("best number:", best_number)
print(f"distance: {best_distance:.2f}")



```

## Optimization template

```python
best_solution = None
best_score = float("inf")

f = lambda x: (x-12)**2 + 8

def evaluate(candidate):
    return abs(15-f(candidate))

possible = np.linspace(-100, 100, 1_000_000)

for candidate in possible:
    score = evaluate(candidate)
    if score < best_score:   # or ">' for maximum problems
        best_score = score
        best_solution = candidate

print(best_solution, f(best_solution), best_score)



```

```python
best_solution = None
best_score = float("inf")

T = lambda x: (x**2) - (10*x) + 30

def evaluate(candidate):
    return T(candidate)

possible = np.linspace(0, 20, 1_000_000)

for candidate in possible:
    score = evaluate(candidate)
    if score < best_score:   # or ">' for maximum problems
        best_score = score
        best_solution = candidate

print(best_solution, T(best_solution), best_score)
```

```python
best_x = None
best_y = None
best_score = float("inf")

E = lambda x, y: ((x-4)**2) + ((y+2)**2)

def evaluate(x, y):
    return E(x, y)

possible_x = np.linspace(-10, 10, 10_000)
possible_y = np.linspace(-10, 10, 10_000)

for x in possible_x:
    for y in possible_y:
        score = evaluate(x, y)
        if score < best_score:   # or ">' for maximum problems
            best_score = score
            best_x = x
            best_y = y

print(f"({best_x}, {best_y})", E(best_x, best_y), best_score)
```

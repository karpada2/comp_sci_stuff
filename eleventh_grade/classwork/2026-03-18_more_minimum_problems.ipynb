```python
import math

step_size = 0.1
allowed_reductions = 4

f = lambda x: math.sin(x) + 0.2*x

current_best_solution = 15
current_best_value = f(current_best_solution)

visited = []

while allowed_reductions > 0:
    visited.append(current_best_solution)
    value_increased = f(current_best_solution + step_size)
    value_decreased = f(current_best_solution - step_size)
    if (value_increased > current_best_value and value_decreased > current_best_value):
        step_size *= 0.1
        allowed_reductions -= 1
    elif value_increased < current_best_value and value_decreased > current_best_value:
        current_best_solution += step_size
    elif value_increased > current_best_value and value_decreased < current_best_value:
        current_best_solution -= step_size
    else:
        if value_increased > value_decreased:
            current_best_solution -= step_size
        else:
            current_best_solution += step_size
    
    if (current_best_solution in visited):
        step_size *= 0.1
        allowed_reductions -= 1

print(current_best_solution)
```

```python
def calculate_score(inpt):
    score = 0
    for i in range(len(inpt)-1):
        score += abs(inpt[i] - inpt[i+1])
    return score

numbers = [3, 1, 4, 7, 2]

for _ in range(2):
    for i in range(len(numbers)):
        do_nothing_score = calculate_score(numbers)
        swap_left = numbers.copy()
        if i > 0:
            temp = swap_left[i]
            swap_left[i] = swap_left[i-1]
            swap_left[i-1] = temp
        swap_right = numbers.copy()
        if i < len(numbers)-1:
            temp = swap_right[i]
            swap_right[i] = swap_right[i-1]
            swap_right[i-1] = temp
        
        swap_left_score = calculate_score(swap_left)
        swap_right_score = calculate_score(swap_right)

        if (swap_left_score == swap_right_score and swap_left_score < do_nothing_score):
            numbers = swap_left
        elif (swap_left_score < swap_right_score and swap_left_score < do_nothing_score):
            numbers = swap_left
        elif (swap_right_score < swap_left_score and swap_right_score < do_nothing_score):
            numbers = swap_right


print(numbers)
```

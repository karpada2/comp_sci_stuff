```python
observation = [
    ("A",16,18),
    ("B",3,5),
    ("C",0,6),
    ("D",1,4),
    ("E",18,20),
    ("F",5,9),
    ("G",8,12),
    ("H",8,11),
    ("I",6,10),
    ("J",2,14),
    ("K",12,16),
    ("L",13,17),
    ("M",5,7),
    ("N",17,19),
    ("O",3,9)
]

def is_valid(combination):
    for i in range(len(combination)):
        for j in range(len(combination)):
                if (combination[j][1] < combination[i][2] and combination[j][1] > combination[i][1]) or (combination[j][2] > combination[i][1] and combination[j][2] < combination[i][2]):
                    return False
            
    return True

def get_combination(take_observation):
    combination = []
    for i in range(len(observation)):
        if take_observation[i]:
            combination.append(observation[i])
    
    if is_valid(combination):
        return combination
    else:
        return []
    

optimal_combination = []
take_observation_og = [False for _ in observation]
def get_optimal(take_observation, index):
    if (index == len(take_observation)):
        return get_combination(take_observation)
        
    take_observation_true = take_observation.copy()
    take_observation_true[index] = True
    take_observation_false = take_observation.copy()
    take_observation_false[index] = False
    
    optimal_true = get_optimal(take_observation_true, index+1)
    optimal_false = get_optimal(take_observation_false, index+1)
    if len(optimal_true) > len(optimal_false):
        return optimal_true
    else:
        return optimal_false

print(get_optimal(take_observation_og, 0))
observation.sort(key=lambda x: x[2])
```
